```python
import torch
from diffusers.models import AutoencoderKL

# Initialize the VAE
vae = AutoencoderKL.from_pretrained("stabilityai/sd-vae-ft-mse")

# Create random input
x = torch.randn(1, 3, 512, 512)  # [batch_size, channels, height, width]

# Encode the input properly
latent = vae.encode(x).latent_dist.sample()
print("Latent shape:", latent.shape)  # Should be [1, 4, 64, 64]

# Decode the latent representation
decoded = vae.decode(latent).sample
print("Decoded shape:", decoded.shape)  # Should be [1, 3, 512, 512]
```
