```python
import torch
from diffusers.models import AutoencoderKL
from diffusers import StableDiffusionPipeline
```

```python
vae = AutoencoderKL.from_pretrained("stabilityai/sd-vae-ft-mse")
encoder = vae.encoder
decoder = vae.decoder
```

```python
x = torch.randn(1, 3, 512, 512)
z = encoder(x)
z.shape
```

```python
z_latent = z[:, :4, :, :]  # Take only the first 4 channels
x_recon = decoder(z_latent)
print(x_recon.shape)
```
